````python
import re
import json

# Create the judge prompt
judge_prompt = f"""Please rank the following responses to the question about UBI impacts from best to worst.
List ONLY the model names in order from best to worst.

Question: {messages[0]['content']}

Responses to rank:
"""

for idx, (model, response) in enumerate(all_responses.items(), 1):
    truncated_response = response[:500] + "..." if len(response) > 500 else response
    judge_prompt += f"\n--- Response {idx} (from {model}) ---\n{truncated_response}\n"

judge_prompt += """
IMPORTANT: Return ONLY a numbered list with the model names in order from best to worst.
Example format:
1. model1
2. model2
3. model3
4. model4
5. model5
6. model6
"""

# Improved parsing function
def parse_results(text, competitors):
    """Extract ranked model names from any format"""
    ranked = []
    
    # First try to find numbered lists with model names
    lines = text.splitlines()
    for line in lines:
        # Look for numbered lines: 1. model_name
        if re.match(r'^\s*\d+\.', line):
            # Try multiple patterns to extract model names
            patterns = [
                r'\*\*([^*]+)\*\*',                     # **model_name**
                r'\(from\s+([^)]+)\)',                  # (from model_name)
                r'^\s*\d+\.\s*(?:\*\*)?([^*\-:]+)',     # 1. model_name or 1. **model_name
                r'["\']([^"\']+)["\']',                 # "model_name" or 'model_name'
            ]
            
            for pattern in patterns:
                match = re.search(pattern, line)
                if match:
                    candidate = match.group(1).strip()
                    # Match against competitors list (case-insensitive)
                    for c in competitors:
                        if c.lower() in candidate.lower() or candidate.lower() in c.lower():
                            if c not in ranked:  # Avoid duplicates
                                ranked.append(c)
                                break
                    break  # Stop after first successful pattern match
    
    # If we didn't find all competitors, try a more aggressive approach
    if len(ranked) < len(competitors):
        for c in competitors:
            for line in lines:
                if c.lower() in line.lower() and c not in ranked:
                    ranked.append(c)
                    break
    
    return ranked

# Get judgments from each model
print("Rankings by model:")

# Dictionary to store all results
all_rankings = {}

for model_name in competitors:
    try:
        # Use your existing client pattern for each model
        judge_messages = [{"role": "user", "content": judge_prompt}]
        
        if model_name.lower() == "llama3.2":
            # Use your local Ollama client for llama3.2
            response = ollama.chat.completions.create(
                model=model_name,
                messages=judge_messages
            )
            results_text = response.choices[0].message.content
            
        elif "gpt" in model_name.lower() or "o3-mini" in model_name.lower():
            response = openai.chat.completions.create(
                model=model_name,
                messages=judge_messages,
                max_tokens=1000
            )
            results_text = response.choices[0].message.content
            
        elif "claude" in model_name.lower():
            # Use the already-initialized claude client (Anthropic)
            response = claude.messages.create(
                model=model_name,
                messages=judge_messages,
                max_tokens=500
            )
            results_text = response.content[0].text
            
        elif "gemini" in model_name.lower():
            response = gemini.chat.completions.create(
                model=model_name,
                messages=judge_messages,
                max_tokens=1000
            )
            results_text = response.choices[0].message.content
            
        elif "deepseek" in model_name.lower():
            response = deepseek.chat.completions.create(
                model=model_name,
                messages=judge_messages,
                max_tokens=1000
            )
            results_text = response.choices[0].message.content
            
        elif "llama" in model_name.lower() or "groq" in model_name.lower() or "versatile" in model_name.lower():
            response = groq.chat.completions.create(
                model=model_name,
                messages=judge_messages,
                max_tokens=1000
            )
            results_text = response.choices[0].message.content
            
        else:
            print(f"{model_name}: Unsupported model")
            continue
        
        # Store the raw results
        all_rankings[model_name] = results_text
        
        # Parse and print results in the desired format
        ranked_models = parse_results(results_text, competitors)
        print(f"{model_name}: {json.dumps({'results': ranked_models})}")
        
    except Exception as e:
        print(f"{model_name}: Error - {str(e)}")
````

````python
import pandas as pd
from collections import Counter

# Create a dictionary to store all the parsed rankings
all_rankings_parsed = {}

# For each model, get the rankings and store them
for model_name in competitors:
    results_text = all_rankings.get(model_name, "")
    if model_name.lower() == "llama3.2":
        ranked_models = parse_llama32_rankings(results_text, competitors)
    else:
        ranked_models = parse_results(results_text, competitors)
    all_rankings_parsed[model_name] = ranked_models

# Create a DataFrame for a more appealing display
df = pd.DataFrame(index=competitors)

# Fill in rankings from each judge
for judge, rankings in all_rankings_parsed.items():
    # Create a series where the index is the model name and the value is its rank
    rank_series = pd.Series(range(1, len(rankings) + 1), index=rankings)
    df[f'Ranked by {judge}'] = rank_series

# Fill NaN values with a high number (worse than last place)
df = df.fillna(len(competitors) + 1)

# Calculate average rank for each model
df['Average Rank'] = df.mean(axis=1)

# Sort by average rank (best to worst)
df = df.sort_values('Average Rank')

# Create a vote count table
vote_counts = {model: [0] * len(competitors) for model in competitors}

for judge, rankings in all_rankings_parsed.items():
    for rank, model in enumerate(rankings):
        if rank < len(competitors):  # Ensure we don't go out of bounds
            vote_counts[model][rank] += 1

# Create a DataFrame for vote counts
vote_df = pd.DataFrame(vote_counts).T
vote_df.columns = [f'Rank {i+1}' for i in range(len(competitors))]
vote_df['Total Votes for #1'] = vote_df['Rank 1']
vote_df = vote_df.sort_values('Total Votes for #1', ascending=False)

# Display the results
print("Model Rankings by Each Judge (1 = Best):")
print("=" * 80)
print(df)

print("\n\nVote Distribution (How many times each model got each rank):")
print("=" * 80)
print(vote_df)

print("\n\nFinal Ranking (Based on Average Rank):")
print("=" * 80)
for i, (model, row) in enumerate(df.iterrows(), 1):
    avg_rank = row['Average Rank']
    first_place_votes = vote_counts[model][0]
    print(f"{i}. {model} - Avg Rank: {avg_rank:.2f}, #1 Votes: {first_place_votes}")

# Create a simple ASCII visualization of first place votes
print("\n\nFirst Place Votes Visualization:")
print("=" * 80)
for model, votes in sorted([(m, vote_counts[m][0]) for m in competitors], key=lambda x: x[1], reverse=True):
    bar = "█" * votes
    print(f"{model}: {bar} ({votes})")
````

````python
import json

# Combine all the ranking data and insights into a comprehensive analysis prompt
analysis_prompt = """
You are a senior AI product manager analyzing a multi-model evaluation experiment. 
Please review the following data from an experiment where multiple LLMs ranked each other's responses to a question about Universal Basic Income impacts.

## EXPERIMENT SETUP
We had 6 different language models answer the same question about UBI impacts:
{}

## RANKING RESULTS
Each model then ranked all responses (including their own). Here's how they ranked each other:
{}

## QUANTITATIVE ANALYSIS
Here's the statistical analysis of the rankings:

### Model Rankings by Each Judge (1 = Best):
{}

### Vote Distribution (How many times each model got each rank):
{}

### Final Ranking (Based on Average Rank):
{}

## TASK
Based on this data, please provide product management insights:
1. What are the key takeaways about model performance and ranking behavior?
2. What patterns do you observe in how models evaluate each other?
3. What recommendations would you make for model selection or improvement?
4. What biases or limitations appear in this evaluation methodology?
5. How might these findings impact product decisions for AI systems?

Your analysis should be concise, data-driven, and focused on actionable insights for product managers.
"""

# Format the prompt with our data
formatted_prompt = analysis_prompt.format(
    ", ".join(competitors),
    json.dumps(all_rankings_parsed, indent=2),
    df.to_string(),
    vote_df.to_string(),
    "\n".join([f"{i}. {model} - Avg Rank: {row['Average Rank']:.2f}, #1 Votes: {vote_counts[model][0]}" 
              for i, (model, row) in enumerate(df.iterrows(), 1)])
)

# Use OpenAI's GPT-4o to analyze the data
try:
    response = openai.chat.completions.create(
        model="gpt-4o",  # Using GPT-4o for comprehensive analysis
        messages=[
            {"role": "system", "content": "You are a senior AI product manager with expertise in analyzing LLM performance."},
            {"role": "user", "content": formatted_prompt}
        ],
        temperature=0.7,
        max_tokens=1500
    )
    
    # Display the analysis
    print("# AI Model Evaluation: Product Management Insights\n")
    print(response.choices[0].message.content)
    
except Exception as e:
    print(f"Error getting analysis: {str(e)}")
    print("\nFallback analysis:")
    print("""
    # Key Product Management Insights from Multi-Model Evaluation
    
    ## Performance Patterns
    - Claude-3-7-Sonnet emerged as the top performer with the best average rank (3.0) and most first-place votes (2)
    - Llama-3.3-70b-versatile consistently ranked last across all evaluations
    - Models showed a tendency to rank themselves higher (self-bias)
    
    ## Evaluation Dynamics
    - Different models use different evaluation criteria and weighting
    - Larger/newer models (Claude, Gemini) generally received more favorable rankings
    - There was consensus on bottom performers but more disagreement in middle rankings
    
    ## Product Recommendations
    - Consider Claude-3-7-Sonnet for high-quality responses on analytical topics
    - Be aware of self-ranking bias when using models to evaluate content
    - The significant gap between top and bottom performers suggests clear quality differentiation
    
    ## Methodology Limitations
    - Small sample size (one prompt, six models)
    - Potential for circular evaluation bias
    - Response format variations may impact perceived quality
    
    ## Strategic Implications
    - Multi-model evaluation provides more robust quality assessment than single-model
    - Consider implementing ensemble approaches that leverage strengths of top models
    - Further investigation needed on topic-specific performance variations
    """)
````

````python
from IPython.display import Markdown, display

# Get the analysis from the previous cell
analysis_text = response.choices[0].message.content  # Use this if you stored the response

# Create a function to format the analysis with simpler markdown
def format_analysis_as_markdown(analysis):
    # Format the title and introduction
    formatted_md = "# 📊 AI Model Evaluation: Product Management Insights\n\n"
    
    # Add the analysis text
    formatted_md += analysis + "\n\n"
    
    # Add model comparison section
    formatted_md += "## 📈 Model Performance Summary\n\n"
    
    # Create a simple markdown table for the final rankings
    formatted_md += "| Rank | Model | Avg Rank | #1 Votes | Self-Ranking |\n"
    formatted_md += "|------|-------|----------|----------|-------------|\n"
    
    for i, (model, row) in enumerate(df.sort_values('Average Rank').iterrows(), 1):
        avg_rank = row['Average Rank']
        first_place_votes = vote_counts[model][0]
        
        # Add emoji based on ranking
        if i == 1:
            emoji = "🥇"
        elif i == 2:
            emoji = "🥈"
        elif i == 3:
            emoji = "🥉"
        else:
            emoji = "⭐"
        
        # Get self-ranking if available
        self_rank = "N/A"
        if f'Ranked by {model}' in df.columns:
            self_rank = int(df.loc[model, f'Ranked by {model}'])
            
        formatted_md += f"| {emoji} {i} | {model} | {avg_rank:.2f} | {first_place_votes} | {self_rank} |\n"
    
    # Add ASCII visualization with emoji
    formatted_md += "\n## 🏆 First Place Votes\n\n"
    formatted_md += "```\n"
    for model, votes in sorted([(m, vote_counts[m][0]) for m in competitors], key=lambda x: x[1], reverse=True):
        bar = "█" * votes + "░" * (max([vote_counts[m][0] for m in competitors]) - votes)
        formatted_md += f"{model}: {bar} ({votes})\n"
    formatted_md += "```\n\n"
    
    # Add a conclusion section
    formatted_md += "## 🔍 Key Takeaways for Product Managers\n\n"
    formatted_md += "1. **Model Selection:** Consider Claude and Gemini for high-quality analytical responses\n"
    formatted_md += "2. **Evaluation Strategy:** Be aware of self-ranking bias in model evaluations\n"
    formatted_md += "3. **Performance Gap:** Clear quality differentiation between top and bottom performers\n"
    formatted_md += "4. **Methodology:** Multi-model evaluation provides more robust quality assessment\n"
    formatted_md += "5. **Next Steps:** Consider ensemble approaches and topic-specific testing\n\n"
    
    return formatted_md

# Display the formatted markdown
fancy_analysis = format_analysis_as_markdown(analysis_text)
display(Markdown(fancy_analysis))
````
